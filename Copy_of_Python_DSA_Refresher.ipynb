```python
days= float(input("Enter a number of days: "))
#number of seconds = 24h per day, 60m per hour, 60s per min
seconds= days * 24 * 60 * 60
print("The number of seconds in ", days , " day(s) is " ,seconds)
```

```python
import math
radius= float(input("Enter a radius: "))
volume=4/3 * math.pi * (radius ** 3)
print(f"the volume of a sphere with radius {radius} is {volume}.")
```

```python
def square_area(length):
    return length**2
def square_perimeter(length):
    return length*4
length= float(input("Enter a length: "))
area= square_area(length)
perimeter= square_perimeter(length)
print(f"A square of side {length} has an area of {area} and a perimeter of {perimeter}.")
```

```python
def case_check(letter):
    if letter.isupper():
        return "uppercase."
    elif letter.islower():
        return "lowercase."
    else:
        return "not a letter."
letter= input("Enter a character: ")
result= case_check(letter)
print(f"The character {letter} is {result}")
```

```python
x=0
y=20
while y>=6:
    y-=4
    x+=2/y
print("The final answer is", x)
```

```python
numbers=[]
for i in range(5):
    number= float(input(f"Enter value {i+1}: "))
    numbers.append(number)
total= sum(numbers)
average= total/len(numbers)
print(f"The array of numbers was {numbers}.\nThe sum is {total} while the average is {average}.")
```
